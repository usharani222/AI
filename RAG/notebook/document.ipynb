```python
### create a text file

import os 

os.makedirs("../data/text_files",exist_ok=True)
```

```python
sample_texts= {
"../data/text_files/python_intro.txt":

"""
Python Programming Language 
Python is a high-level, interpreted programming language known for its simplicity and readability. It supports multiple programming paradigms, including procedural, object-oriented, and functional programming. Python's extensive standard library and active community make it a popular choice for web development, data analysis, artificial intelligence, scientific computing, and more.
key features of Python include dynamic typing, automatic memory management, and a large ecosystem of third-party packages. It is widely used in various domains, from web development frameworks like Django and Flask to data science libraries such as NumPy, pandas, and TensorFlow. Python's versatility and ease of use have contributed to its widespread adoption across industries and educational institutions.
Python's syntax emphasizes code readability, allowing developers to express concepts in fewer lines of code compared to other languages. Its interactive interpreter and extensive documentation make it accessible for beginners while still being powerful enough for experienced programmers. Overall, Python's combination of simplicity, versatility, and a strong community has solidified its position as one of the most popular programming languages in the world.
""",

"../data/text_files/data_science.txt":
"""
Data Science is an interdisciplinary field that combines statistical analysis, machine learning, data visualization, and domain expertise to extract insights and knowledge from structured and unstructured data. It involves the collection, cleaning, and preprocessing of data, followed by exploratory data analysis to identify patterns and trends. Data scientists use various techniques to build predictive models, perform hypothesis testing, and make data-driven decisions. The field has applications in diverse industries such as finance, healthcare, marketing, and technology, enabling organizations to optimize processes, improve customer experiences, and drive innovation. With the increasing availability of big data and advancements in computing power, data science continues to play a crucial role in shaping the future of business and research.
"""
}

for file_path, content in sample_texts.items():
    with open(file_path, "w") as f:
        f.write(content)

print("Sample text files created successfully.")
```

```python
### TextLoader

from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/python_intro.txt",encoding="utf-8")
document = loader.load()
print(document)
```

```python
### Directory Loader

from langchain_community.document_loaders import DirectoryLoader

loader = DirectoryLoader("../data/text_files",
                         glob="**/*.txt",
                         loader_cls=TextLoader,
                         loader_kwargs={"encoding":"utf-8"}
)
documents=loader.load()
for d in documents:
    print(d.page_content)
    print(d.metadata)
```

```python
## pdfs

from langchain_community.document_loaders import PyPDFLoader,PyMuPDFLoader

loader = DirectoryLoader(
    "../data/pdf",
    loader_cls=PyMuPDFLoader,
    glob="**/*.pdf"
)

documents=loader.load()
documents
```
